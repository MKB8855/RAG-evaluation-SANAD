```python
import re

def count_article_headings(text):
    text = str(text)

    # Count "المادة" only when it appears at the beginning of a line
    headings = re.findall(
        r'(?m)^\s*[\u200f\u200e]*المادة\s+[^\n]{1,50}',
        text
    )

    return len(headings)

df["article_headings"] = df["text"].apply(count_article_headings)

print(df["article_headings"].value_counts().sort_index())

problem_rows = df[df["article_headings"] > 1]

print("Rows containing multiple probable article headings:")
print(len(problem_rows))

display(
    problem_rows[
        ["law_name", "article_number", "article_headings", "text"]
    ].head(20)
)
```

```python
import pandas as pd
import re

# Keep original data safe
df_original = df.copy()

# Pattern: "المادة" appearing at the beginning of a line
heading_pattern = re.compile(
    r'(?m)^[ \t\u200e\u200f]*المادة\s+[^\n]+'
)

new_rows = []

for idx, row in df_original.iterrows():
    text = str(row["text"])
    matches = list(heading_pattern.finditer(text))

    # If there are 0 or 1 headings, keep the row as it is
    if len(matches) <= 1:
        new_row = row.to_dict()
        new_row["original_row"] = idx
        new_row["detected_heading"] = (
            matches[0].group().strip() if matches else None
        )
        new_rows.append(new_row)

    # If there are multiple headings, split them
    else:
        for i, match in enumerate(matches):
            start = match.start()

            if i + 1 < len(matches):
                end = matches[i + 1].start()
            else:
                end = len(text)

            article_text = text[start:end].strip()

            new_row = row.to_dict()

            # Preserve original information
            new_row["original_row"] = idx
            new_row["original_article_number"] = row["article_number"]

            # Heading detected from the actual legal text
            heading_line = match.group().strip()
            new_row["detected_heading"] = heading_line

            # Replace text with the separated article
            new_row["text"] = article_text

            new_rows.append(new_row)

# Create a new dataframe
df_split = pd.DataFrame(new_rows)

print("Original rows:", len(df_original))
print("After splitting:", len(df_split))

display(
    df_split[
        ["law_name", "article_number",
         "detected_heading", "text", "original_row"]
    ].head(20)
)
```

```python
import pandas as pd
import re

df = df_original.copy()

# Remove invisible direction characters only
def normalize_text(text):
    text = str(text)
    text = re.sub(r'[\u200e\u200f\u202a-\u202e]', '', text)
    text = re.sub(r'[ \t]+', ' ', text)
    return text.strip()

# Detect a real article heading
def detect_heading(line):
    line = normalize_text(line)

    if not line.startswith("المادة"):
        return None

    # Ignore references to another article
    reference_words = [
        "من النظام", "منالنظام",
        "من اللائحة", "مناللائحة",
        "وفقاً", "وفقا",
        "بموجب",
        "المشار إليها",
        "الواردة في",
        "المنصوص عليها"
    ]

    if any(x in line for x in reference_words):
        return None

    # Example: المادة السادسة: معيار...
    if ":" in line or "：" in line:
        heading = re.split(r'[:：]', line, maxsplit=1)[0].strip()
        return heading

    # Example: المادة الخامسة عشرة
    if len(line) <= 60:
        return line.strip(" .،؛")

    return None


def split_law(text):
    lines = str(text).splitlines()

    articles = []
    preamble = []

    current_heading = None
    current_lines = []

    for line in lines:

        heading = detect_heading(line)

        if heading:

            if current_heading is not None:
                articles.append({
                    "article_number": current_heading,
                    "text": "\n".join(current_lines).strip()
                })

            current_heading = heading
            current_lines = [normalize_text(line)]

        else:
            if current_heading is not None:
                current_lines.append(normalize_text(line))
            else:
                if line.strip():
                    preamble.append(normalize_text(line))

    if current_heading is not None:
        articles.append({
            "article_number": current_heading,
            "text": "\n".join(current_lines).strip()
        })

    return articles, preamble
```

```python
records = []
preambles = []

for law_name, group in df.groupby("law_name", sort=False):

    # Join all pieces belonging to the same law
    full_text = "\n".join(
        group["text"].astype(str).tolist()
    )

    articles, preamble = split_law(full_text)

    law_type = group["law_type"].iloc[0]
    source = group["source"].iloc[0]

    for article in articles:
        records.append({
            "law_name": law_name,
            "law_type": law_type,
            "article_number": article["article_number"],
            "text": article["text"],
            "source": source
        })

    if preamble:
        preambles.append({
            "law_name": law_name,
            "text": "\n".join(preamble)
        })


df_articles = pd.DataFrame(records)

print("Original rows:", len(df))
print("Detected legal articles:", len(df_articles))
print("Laws:", df_articles["law_name"].nunique())
print("Laws with text before first article:", len(preambles))

display(df_articles.head(20))
```

```python
df_articles["article_key"] = (
    df_articles["article_number"]
    .str.replace(r'[\s\(\)\.:：]', '', regex=True)
)

duplicates = df_articles[
    df_articles.duplicated(
        subset=["law_name", "article_key"],
        keep=False
    )
]

print("Duplicate article rows:", len(duplicates))
print(
    "Duplicate article groups:",
    duplicates.groupby(["law_name", "article_key"]).ngroups
)

display(
    duplicates[
        ["law_name", "article_number", "text"]
    ].head(20)
)
```

```python
import re
import pandas as pd

def normalize_for_compare(text):
    text = str(text)
    text = re.sub(r'[\u200e\u200f\u202a-\u202e]', '', text)
    text = re.sub(r'\s+', ' ', text)
    return text.strip()

df_articles["text_norm"] = df_articles["text"].apply(normalize_for_compare)

results = []

for (law, key), group in df_articles.groupby(["law_name", "article_key"]):

    if len(group) == 1:
        continue

    texts = group["text_norm"].tolist()

    if len(set(texts)) == 1:
        status = "Exact Duplicate"

    elif any(
        a != b and (a in b or b in a)
        for i, a in enumerate(texts)
        for j, b in enumerate(texts)
        if i < j
    ):
        status = "Contained / Partial Duplicate"

    else:
        status = "Needs Manual Review"

    results.append({
        "law_name": law,
        "article_key": key,
        "count": len(group),
        "status": status
    })

duplicate_check = pd.DataFrame(results)

print(duplicate_check["status"].value_counts())
display(duplicate_check)
```

```python
import pandas as pd
from difflib import SequenceMatcher

review_rows = []

manual_groups = duplicate_check[
    duplicate_check["status"] == "Needs Manual Review"
]

for _, item in manual_groups.iterrows():

    law = item["law_name"]
    key = item["article_key"]

    group = df_articles[
        (df_articles["law_name"] == law) &
        (df_articles["article_key"] == key)
    ]

    rows = group.to_dict("records")

    # Most groups contain two versions
    if len(rows) >= 2:

        text1 = rows[0]["text"]
        text2 = rows[1]["text"]

        similarity = SequenceMatcher(
            None,
            normalize_for_compare(text1),
            normalize_for_compare(text2)
        ).ratio()

        review_rows.append({
            "law_name": law,
            "article_number_1": rows[0]["article_number"],
            "text_1": text1,
            "length_1": len(text1),

            "article_number_2": rows[1]["article_number"],
            "text_2": text2,
            "length_2": len(text2),

            "similarity": round(similarity, 3),

            # Fill this manually later:
            "decision": ""
        })

manual_review = pd.DataFrame(review_rows)

display(manual_review.head())

manual_review.to_excel(
    "manual_duplicate_review.xlsx",
    index=False
)

print("Manual review cases:", len(manual_review))
```

```python
# Keys that require manual verification
manual_keys = set(
    zip(
        manual_review["law_name"],
        manual_review["article_number_1"]
            .str.replace(r'[\s\(\)\.:：]', '', regex=True)
    )
)

def is_manual(row):
    return (row["law_name"], row["article_key"]) in manual_keys

df_articles["needs_manual_review"] = df_articles.apply(is_manual, axis=1)

# Keep clean articles only for the first RAG experiment
df_clean = df_articles[
    ~df_articles["needs_manual_review"]
].copy()

print("All detected articles:", len(df_articles))
print("Excluded for manual review:", df_articles["needs_manual_review"].sum())
print("Clean articles available for RAG:", len(df_clean))
```

```python
df_clean = df_clean.sort_values(
    "text",
    key=lambda x: x.str.len(),
    ascending=False
)

df_clean = df_clean.drop_duplicates(
    subset=["law_name", "article_key"],
    keep="first"
)

print("Final clean rows:", len(df_clean))
```

```python
df_clean.to_parquet(
    "saudi_legal_moj_clean.parquet",
    index=False
)

manual_review.to_excel(
    "articles_needing_manual_verification.xlsx",
    index=False
)
print("Saved:", len(df_clean), "rows")
```

```python
df_clean["embedding_text"] = (
    "اسم النظام: " + df_clean["law_name"].astype(str) +
    "\nرقم المادة: " + df_clean["article_number"].astype(str) +
    "\nالنص القانوني: " + df_clean["text"].astype(str)
)

texts = df_clean["embedding_text"].tolist()

print("Number of legal articles:", len(texts))
print("\nExample:\n")
print(texts[0][:700])
```

```python
from sentence_transformers import SentenceTransformer
import numpy as np
import faiss

bge_model = SentenceTransformer(
    "BAAI/bge-m3",
    device="cuda"
)

bge_embeddings = bge_model.encode(
    texts,
    batch_size=8,
    show_progress_bar=True,
    normalize_embeddings=True
)

bge_embeddings = np.asarray(
    bge_embeddings,
    dtype="float32"
)

print("Embedding shape:", bge_embeddings.shape)
```

```python
bge_index = faiss.IndexFlatIP(
    bge_embeddings.shape[1]
)

bge_index.add(bge_embeddings)

print("BGE documents indexed:", bge_index.ntotal)
```

```python
qwen_model = SentenceTransformer(
    "Qwen/Qwen3-Embedding-0.6B",
    device="cuda"
)

qwen_embeddings = qwen_model.encode(
    texts,
    batch_size=8,
    show_progress_bar=True,
    normalize_embeddings=True
)

qwen_embeddings = np.asarray(
    qwen_embeddings,
    dtype="float32"
)

print("Embedding shape:", qwen_embeddings.shape)
```

```python
qwen_index = faiss.IndexFlatIP(
    qwen_embeddings.shape[1]
)

qwen_index.add(qwen_embeddings)

print("Qwen documents indexed:", qwen_index.ntotal)
```

```python
faiss.write_index(
    bge_index,
    "bge_m3_saudi_legal.index"
)

faiss.write_index(
    qwen_index,
    "qwen3_saudi_legal.index"
)

df_clean.to_parquet(
    "saudi_legal_moj_clean.parquet",
    index=False
)

print("All Stage 1 files saved successfully.")
```
